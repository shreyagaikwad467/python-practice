```python
number = [10, 29, 20, 34, 32, 56, 54]

num_2 = []

for i in number:
    num_2.append(i * 2)
print(num_2)
```

```python
number = [10, 29, 20, 34, 32, 56, 54]

a = np.array(number)
print(a*2)

# You don't need to use a for loop to multiply each element by 2.
```
